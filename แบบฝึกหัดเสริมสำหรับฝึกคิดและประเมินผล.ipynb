## **1. เขียนโค้ดให้ Agent ดูดฝุ่นเมื่อห้องสกปรก และย้ายถ้าห้องสะอาด**

```python
class Environment:
    def __init__(self):
        self.status = {'A': 'Dirty', 'B': 'Clean'}  # สถานะเริ่มต้น

    def is_dirty(self, location):
        return self.status[location] == 'Dirty'

    def clean(self, location):
        self.status[location] = 'Clean'


class SimpleReflexVacuumAgent:
    def __init__(self, environment):
        self.environment = environment
        self.location = 'A'

    def perceive(self):
        return (self.location, self.environment.is_dirty(self.location))

    def act(self):
        location, is_dirty = self.perceive()
        if is_dirty:
            action = 'Suck'
            self.environment.clean(location)
            print(f"Action: {action} at {location} - Room cleaned!")
        else:
            if location == 'A':
                action = 'Move Right'
                self.location = 'B'
            else:
                action = 'Move Left'
                self.location = 'A'
            print(f"Action: {action} - Moving to {self.location}")
        return action
```

```python
# --- ทดสอบการทำงาน ---
env = Environment()
agent = SimpleReflexVacuumAgent(env)

for step in range(4):
    print(f"\nStep {step + 1}: A={env.status['A']}, B={env.status['B']}")
    agent.act()

print("\nFinal status:", env.status)
```

## **2. เปลี่ยนสถานะเริ่มต้นให้ห้องทั้งสอง “สกปรก” แล้วทดสอบ Agent**

```python
# --- ทดสอบการทำงาน ---
env = Environment()
agent = SimpleReflexVacuumAgent(env)

for step in range(4):
    print(f"\nStep {step + 1}: A={env.status['A']}, B={env.status['B']}")
    agent.act()

print("\nFinal status:", env.status)
```
